```python
import pandas as pd
```

```python
usage_df = pd.read_csv("../data/raw/ev_charging_session_dataset.csv")
```

```python
usage_df.shape
usage_df.info() 
usage_df.describe(include="all")
```

```python

usage_df.head()
```

```python
usage_df["Date"] = pd.to_datetime(usage_df["Date"])
```

```python
usage_df["Charging_Start"] = pd.to_datetime(
    usage_df["Date"].dt.strftime("%Y-%m-%d") + " " +
    usage_df["Charging_Start_Time"]
)

usage_df["Charging_End"] = pd.to_datetime(
    usage_df["Date"].dt.strftime("%Y-%m-%d") + " " +
    usage_df["Charging_End_Time"]
)
```

```python
usage_df.head()
```

```python
usage_df["Session_Duration_Minutes"] = (
    usage_df["Charging_End"] - usage_df["Charging_Start"]
).dt.total_seconds() / 60
```

```python
usage_df[["Charging_Start", "Charging_End", "Session_Duration_Minutes"]].head()
```

```python
usage_df["Start_Hour"] = usage_df["Charging_Start"].dt.hour
```

```python
usage_df["End_Hour"] = usage_df["Charging_End"].dt.hour
```

```python
usage_df["Day_of_Week"] = usage_df["Date"].dt.day_name()
```

```python
usage_df["Is_Weekend"] = usage_df["Date"].dt.weekday >= 5
```

```python
usage_df["Month"] = usage_df["Date"].dt.month
```

```python
usage_df = usage_df[usage_df["City"] == "Delhi"].copy()
```
