```python
import pandas as pd
```

```python
delhi_df = pd.read_csv("../data/processed/cleaned_delhi_ev_data.csv")
```

```python
print(delhi_df.shape)
print(delhi_df.columns)
delhi_df.head()
```

```python
fe_df = delhi_df.copy()
```

```python
fe_df.drop(columns=["uid", "name", "address"], inplace=True)
```

```python
fe_df[["open", "close"]].dtypes
```

```python
fe_df["open"] = pd.to_datetime(fe_df["open"], format="mixed")
fe_df["close"] = pd.to_datetime(fe_df["close"], format="mixed")
```

```python
fe_df[["open", "close"]].dtypes
```

```python
fe_df["open_hour"] = fe_df["open"].dt.hour
fe_df["close_hour"] = fe_df["close"].dt.hour
```

```python
fe_df.drop(columns=["open", "close"], inplace=True)
```

```python
fe_df["supports_2W"] = fe_df["vehicle_type"].str.contains("2W").astype(int)
fe_df["supports_3W"] = fe_df["vehicle_type"].str.contains("3W").astype(int)
fe_df["supports_4W"] = fe_df["vehicle_type"].str.contains("4W").astype(int)
```

```python
fe_df.head()
```

```python
fe_df.drop(columns=["vehicle_type"], inplace=True)
```

```python
fe_df.select_dtypes(include="object").nunique().sort_values()
```

```python
fe_df = pd.get_dummies(
    fe_df,
    columns=["power_type", "zone", "charger_type", "vendor_name"],
    dtype=int
)
```

```python
fe_df.shape
fe_df.head()
fe_df.info()
```

```python
fe_df.select_dtypes(include="object").columns
```

```python
fe_df.to_csv("../data/processed/feature_engineered_ev_dataset.csv", index=False)
```
